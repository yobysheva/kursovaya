```python
import matplotlib.pyplot as plt
import numpy as np
```

# Задаю функцию для вычисления x на текущем шаге

```python
def f(x, xc):
    return 1.0 / (1.0 + np.exp(xc - x))
def m(x, N):
    return x / (N + x)
def Xn(x, r, N, xc = 5):
    F = f(x, xc)
    M = m(x, N)
    return (1 - F + F * M) * r * x
```

# Функция для построения графиков для выявления зависимости  $x(r), r \in [0.01, 5]$ при фиксированных $x_c, N$

```python
def r_bifurcation(ax, xc_val, N_val):
    P = np.linspace(0.01, 15, 2000)
    N0 = N_val
    xc = xc_val
    R = []
    X = []
    for r in P:
        x = np.random.random()
        for n in range(2000):
            x = Xn(x, r, N0, xc)
        for n in range(50):
            x = Xn(x, r, N0, xc)
            R.append(r)
            X.append(x)

    ax.plot(R, X, ls='', marker=',')
    ax.set_title(f'$x_c={xc_val},\\;N={N_val}$')
    ax.set_xlabel('$r$')
    ax.set_ylabel('$x$')
```

```python
fig, axes = plt.subplots(2, 5, figsize=(16, 8))

xc_values = [1.5, 2.5]
N_values  = [10, 30, 100, 300, 500]

for i, xc in enumerate(xc_values):
    for j, N in enumerate(N_values):
        r_bifurcation(axes[i, j], xc, N)

plt.tight_layout()
plt.show()
```

# Функция для построения графиков для выявления зависимости  $x(N), N > 0$ при фиксированных $x_c, r$

```python
def N_bifurcation(ax, xc_val, r_val = 1.5):
    P = np.linspace(0, 400, 800)
    r = r_val
    xc = xc_val
    N = []
    X = []
    for N0 in P:
        x = 1
        for n in range(3025):
            x = Xn(x, r, N0, xc)
        for n in range(10):
            x = Xn(x, r, N0, xc)
            N.append(N0)
            X.append(x)
    ax.plot(N, X, ls='', marker=',')
    ax.set_title(f'$x_c={xc_val},\\; r={r_val}$')
    ax.set_xlabel('$N$')
    ax.set_ylabel('$x$')
```

```python
def r_bifurcation1(xc_val, N_val):
    P = np.linspace(0.01, 5, 5000)
    N0 = N_val
    xc = xc_val
    R = []
    X = []
    for r in P:
        x = np.random.random()
        for n in range(225):
            x = Xn(x, r, N0, xc)
        for n in range(10):
            x = Xn(x, r, N0, xc)
            R.append(r)
            X.append(x)
    plt.plot(R, X, ls='', marker=',')
    plt.title(f'Bifurcation x_c = {xc_val}, N = {N_val}')
    plt.xlabel('r')
    plt.ylabel('x')
    plt.show()
```
